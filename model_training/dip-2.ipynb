```python
# 1. Install YOLO
!pip install ultralytics -q
import yaml
from ultralytics import YOLO

# ==========================================
# PASTE YOUR EXACT KAGGLE FOLDER PATHS HERE:
# ==========================================
train_path = '/kaggle/input/datasets/aliabdelmenam/rdd-2022/RDD_SPLIT/train/images'
val_path   = '/kaggle/input/datasets/aliabdelmenam/rdd-2022/RDD_SPLIT/val/images'

# 2. CREATE THE MISSING YAML FILE AUTOMATICALLY
print("Creating data.yaml file...")
yaml_content = {
    'train': train_path,
    'val': val_path,
    'nc': 4,
    'names': ['D00', 'D10', 'D20', 'D40'] # Class 3 is D40 (Pothole)
}

yaml_file_path = '/kaggle/working/rdd2022_data.yaml'
with open(yaml_file_path, 'w') as f:
    yaml.dump(yaml_content, f)

# 3. LOAD THE NANO MODEL (For Real-Time Mobile Speed)
print("Loading YOLOv8-Nano...")
model = YOLO('yolov8n.pt') 

# 4. TRAIN THE MODEL USING OUR NEW YAML FILE
print("Starting Training...")
model.train(
    data=yaml_file_path, 
    epochs=50,       
    imgsz=640,       
    batch=16         
)

# 5. EXPORT TO ANDROID FORMAT (FLOAT16)
print("\nExporting to Float16 for Android...")
best_model = YOLO('/kaggle/working/runs/detect/train/weights/best.pt')
export_path = best_model.export(format='tflite', half=True)

print(f"\nSUCCESS! Download your 6MB model from: {export_path}")
```
